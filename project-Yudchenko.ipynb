## Введение

Проект посвящён анализу многолетней динамики температуры воздуха и атмосферных осадков в Краснодаре.

В работе используются данные о средних месячных и годовых температурах воздуха за 1881–2026 гг., а также о месячных и годовых суммах осадков за 1938–2026 гг. Источник данных — сайт «Погода и климат».

Цель проекта — оценить изменение температурного режима и количества осадков в Краснодаре за длительный период, а также сравнить климатические характеристики двух 30-летних периодов.

```python
import numpy as np
import matplotlib.pyplot as plt
temperature_data = np.loadtxt(
    'data/krasnodar_temperature.csv',
    delimiter=',',
    skiprows=1
)
precipitation_data = np.loadtxt(
    'data/krasnodar_precipitation.csv',
    delimiter=',',
    skiprows=1
)
```

## Подготовка данных

В исходном наборе данных отсутствующие значения обозначены числом 999.9 для температур и -999.9 для осадков. Чтобы это значение не влияло на расчёты средних, минимальных и максимальных температур и осадков, нужно заменить его на NaN.

```python
temperature_mask = temperature_data == 999.9
temperature_data[temperature_mask] = np.nan

precipitation_mask = precipitation_data == -999.9
precipitation_data[precipitation_mask] = np.nan
```

## Годовая температура воздуха

Для анализа многолетней динамики температуры выделим значения средней годовой температуры воздуха.

```python
years = temperature_data[:, 0]
annual_temperature = temperature_data[:, -1]
mean_annual_temperature = np.nanmean(annual_temperature)
min_annual_temperature = np.nanmin(annual_temperature)
max_annual_temperature = np.nanmax(annual_temperature)
min_index = np.nanargmin(annual_temperature)
max_index = np.nanargmax(annual_temperature)
min_year = years[min_index]
max_year = years[max_index]
round_mean_annual_temperature = round(mean_annual_temperature, 1)
round_min_annual_temperature = round(min_annual_temperature, 1)
round_max_annual_temperature = round(max_annual_temperature, 1)
print(f'Средняя годовая температура за весь период: {round_mean_annual_temperature} °C')
print(f'Минимальная среднегодовая температура: {round_min_annual_temperature} °C, наблюдалась в {int(min_year)} году.')
print(f'Максимальная среднегодовая температура: {round_max_annual_temperature} °C, наблюдалась в {int(max_year)} году.')
```

## Динамика годовой температуры

На графике заметны значительные межгодовые колебания температуры.

```python
plt.figure(figsize=(14, 7))
plt.plot(years, annual_temperature, label='Среднегодовая температура')
plt.axhline(mean_annual_temperature, linestyle='--', label='Средняя годовая температура за весь период')
plt.xlabel('Год')
plt.ylabel('Температура, °C')
plt.title('Динамика среднегодовой температуры воздуха в Краснодаре')
plt.legend()
plt.show()
```

## Температура января

Для оценки изменения зимнего температурного режима отдельно рассмотрим среднюю температуру января.

```python
january_temperature = temperature_data[:, 1]
mean_january_temperature = np.nanmean(january_temperature)
min_january_temperature = np.nanmin(january_temperature)
max_january_temperature = np.nanmax(january_temperature)
min_january_index = np.nanargmin(january_temperature)
max_january_index = np.nanargmax(january_temperature)
min_january_year = years[min_january_index]
max_january_year = years[max_january_index]
round_mean_january_temperature = round(mean_january_temperature, 1)
round_min_january_temperature = round(min_january_temperature, 1)
round_max_january_temperature = round(max_january_temperature, 1)
print(f'Средняя температура января за весь период: {round_mean_january_temperature} °C')
print(f'Минимальная средняя температура января: {round_min_january_temperature} °C, наблюдалась в {int(min_january_year)} году.')
print(f'Максимальная средняя температура января: {round_max_january_temperature} °C, наблюдалась в {int(max_january_year)} году.')
```

## Динамика температуры января
Средняя температура января в Краснодаре характеризуется значительными межгодовыми колебаниями. В начале периода чаще наблюдались отрицательные значения температуры, а в последние десятилетия наблюдаются январи с температурой выше 0 °C. В целом график указывает на повышение зимних температур к концу рассматриваемого периода.

```python
plt.figure(figsize=(14, 7))
plt.plot(years, january_temperature, label='Средняя температура января')
plt.axhline(mean_january_temperature, linestyle='--', label='Средняя температура января за весь период')
plt.xlabel('Год')
plt.ylabel('Температура, °C')
plt.title('Динамика средней температуры января в Краснодаре')
plt.legend()
plt.show()
```

## Температура июля

Для оценки изменения летнего температурного режима отдельно рассмотрим среднюю температуру июля.

```python
july_temperature = temperature_data[:, 7]
mean_july_temperature = np.nanmean(july_temperature)
min_july_temperature = np.nanmin(july_temperature)
max_july_temperature = np.nanmax(july_temperature)
min_july_index = np.nanargmin(july_temperature)
max_july_index = np.nanargmax(july_temperature)
min_july_year = years[min_july_index]
max_july_year = years[max_july_index]
round_mean_july_temperature = round(mean_july_temperature, 1)
round_min_july_temperature = round(min_july_temperature, 1)
round_max_july_temperature = round(max_july_temperature, 1)
print(f'Средняя температура июля за весь период: {round_mean_july_temperature} °C')
print(f'Минимальная средняя температура июля: {round_min_july_temperature} °C, наблюдалась в {int(min_july_year)} году.')
print(f'Максимальная средняя температура июля: {round_max_july_temperature} °C, наблюдалась в {int(max_july_year)} году.')
```

## Динамика температуры июля

Также заметны скачки, однако средняя температура июля меняется не так интенсивно, как температура января. Видно общее повышение летних температур, особенно после 1980 года, значения выше 25 °C стали встречаться чаще.

```python
plt.figure(figsize=(14, 7))
plt.plot(years, july_temperature, label='Средняя температура июля')
plt.axhline(mean_july_temperature, linestyle='--', label='Среднее за весь период')
plt.xlabel('Год')
plt.ylabel('Температура, °C')
plt.title('Динамика средней температуры июля в Краснодаре')
plt.legend()
plt.show()
```

## Годовое количество осадков

Для оценки изменения режима влаги рассмотрим годовое количество осадков. Годы с пропусками месячных значений в расчете участвовать не будут.

```python
precipitation_years = precipitation_data[:, 0]
annual_precipitation = precipitation_data[:, -1]
precipitation_mask = ~np.isnan(annual_precipitation)
precipitation_years_clean = precipitation_years[precipitation_mask]
annual_precipitation_clean = annual_precipitation[precipitation_mask]
mean_annual_precipitation = np.mean(annual_precipitation_clean)
min_annual_precipitation = np.min(annual_precipitation_clean)
max_annual_precipitation = np.max(annual_precipitation_clean)
min_precipitation_index = np.argmin(annual_precipitation_clean)
max_precipitation_index = np.argmax(annual_precipitation_clean)
min_precipitation_year = precipitation_years_clean[min_precipitation_index]
max_precipitation_year = precipitation_years_clean[max_precipitation_index]
round_mean_annual_precipitation = round(mean_annual_precipitation, 1)
round_min_annual_precipitation = round(min_annual_precipitation, 1)
round_max_annual_precipitation = round(max_annual_precipitation, 1)
print(f'Средняя годовая сумма осадков за весь период: {round_mean_annual_precipitation} мм')
print(f'Минимальная годовая сумма осадков: {round_min_annual_precipitation} мм, наблюдалась в {int(min_precipitation_year)} году.')
print(f'Максимальная годовая сумма осадков: {round_max_annual_precipitation} мм, наблюдалась в {int(max_precipitation_year)} году.')
```

## Динамика осадков

Годовая сумма осадков в Краснодаре характеризуется сильной межгодовой изменчивостью. Большинство значений находится примерно в диапазоне 500–850 мм, при среднем уровне 690.2 мм в год. Выраженного устойчивого роста или снижения количества осадков за весь период по графику не видно, однако отдельные очень влажные и сухие годы встречаются на всём протяжении наблюдений, хотя периоды длительной засухи характерны больше для начала.

```python
plt.figure(figsize=(14, 7))
plt.plot(precipitation_years, annual_precipitation, label='Годовая сумма осадков')
plt.axhline(mean_annual_precipitation, linestyle='--', label='Среднее за весь период')
plt.xlabel('Год')
plt.ylabel('Осадки, мм')
plt.title('Динамика годовой суммы осадков в Краснодаре')
plt.legend()
plt.show()
```

## Климатограмма Краснодара за 1945–1974 гг.

Для характеристики климата первого периода рассчитаем среднемесячную температуру воздуха и среднее месячное количество осадков за 1945–1974 гг.


```python
temperature_years = temperature_data[:, 0]
precipitation_years = precipitation_data[:, 0]
first_temperature_mask = (temperature_years >= 1945) & (temperature_years <= 1974)
first_precipitation_mask = (precipitation_years >= 1945) & (precipitation_years <= 1974)
first_temperature = np.nanmean(temperature_data[first_temperature_mask, 1:13], axis=0)
first_precipitation = np.nanmean(precipitation_data[first_precipitation_mask, 1:13], axis=0)
months = ['Янв', 'Фев', 'Мар', 'Апр', 'Май', 'Июн', 'Июл', 'Авг', 'Сен', 'Окт', 'Ноя', 'Дек']
fig, ax1 = plt.subplots(figsize=(12, 6))
ax2 = ax1.twinx()
ax1.bar(months, first_precipitation, label='Осадки')
ax2.plot(months, first_temperature, marker='o', color='orange', label='Температура')
ax1.set_xlabel('Месяц')
ax1.set_ylabel('Осадки, мм')
ax2.set_ylabel('Температура, °C')
plt.title('Климатограмма Краснодара, 1945–1974 гг.')
plt.show()
```

## Климатограмма Краснодара за 1996–2025 гг.

Для характеристики современного периода рассчитаем среднемесячную температуру воздуха и среднее месячное количество осадков за 1996–2025 гг.

```python
second_temperature_mask = (temperature_years >= 1996) & (temperature_years <= 2025)
second_precipitation_mask = (precipitation_years >= 1996) & (precipitation_years <= 2025)
second_temperature = np.nanmean(temperature_data[second_temperature_mask, 1:13], axis=0)
second_precipitation = np.nanmean(precipitation_data[second_precipitation_mask, 1:13], axis=0)
fig, ax1 = plt.subplots(figsize=(12, 6))
ax2 = ax1.twinx()
ax1.bar(months, second_precipitation, label='Осадки')
ax2.plot(months, second_temperature, marker='o', color='orange', label='Температура')
ax1.set_xlabel('Месяц')
ax1.set_ylabel('Осадки, мм')
ax2.set_ylabel('Температура, °C')
plt.title('Климатограмма Краснодара, 1996–2025 гг.')
plt.show()
```

## Сравнение климатограмм
Сравнение двух климатограмм показывает заметное потепление. Среднегодовая температура во втором периоде выше на 1,8 °C. Прослеживается небольшое сглаживание сезонности, так как температура января увеличилась сильнее (на 2,4 °C), чем температура июля (на 2 °C). 

Изменения режима осадков выражены слабее. Средняя годовая сумма осадков во втором периоде увеличилась на 56,9 мм (с учетом пропуска одного года во второй период, если экстраполировать значения может будет около 60 мм). При этом общий сезонный ход осадков в основном сохранился, только август стал суше, а доля осадков приходящаяся на сентябрь наоборот повысилась. 

```python
first_mean_temperature = np.nanmean(temperature_data[first_temperature_mask, 13])
second_mean_temperature = np.nanmean(temperature_data[second_temperature_mask, 13])
temperature_difference = second_mean_temperature - first_mean_temperature
first_january_temperature = np.nanmean(temperature_data[first_temperature_mask, 1])
second_january_temperature = np.nanmean(temperature_data[second_temperature_mask, 1])
january_difference = second_january_temperature - first_january_temperature
first_july_temperature = np.nanmean(temperature_data[first_temperature_mask, 7])
second_july_temperature = np.nanmean(temperature_data[second_temperature_mask, 7])
july_difference = second_july_temperature - first_july_temperature
first_mean_precipitation = np.nanmean(precipitation_data[first_precipitation_mask, 13])
second_mean_precipitation = np.nanmean(precipitation_data[second_precipitation_mask, 13])
precipitation_difference = second_mean_precipitation - first_mean_precipitation
round_temperature_difference = round(temperature_difference, 1)
round_january_difference = round(january_difference, 1)
round_july_difference = round(july_difference, 1)
round_precipitation_difference = round(precipitation_difference, 1)
print(f'Изменение среднегодовой температуры: {round_temperature_difference} °C')
print(f'Изменение средней температуры января: {round_january_difference} °C')
print(f'Изменение средней температуры июля: {round_july_difference} °C')
print(f'Изменение средней годовой суммы осадков: {round_precipitation_difference} мм')
```

## Выводы

Были рассмотрены среднегодовые значения, январские и июльские температуры, а также годовые суммы осадков. Дополнительно были построены климатограммы для двух 30-летних периодов — 1945–1974 и 1996–2025 гг., чтобы сравнить изменение климатических характеристик во времени. 

Результаты показывают, что наиболее выраженные изменения связаны с ростом температуры воздуха, причём зимние температуры увеличились несколько сильнее летних. Изменения режима осадков выражены слабее и проявляются преимущественно в перераспределении осадков между отдельными месяцами. Это (как общее потепление, так и более быстрый темп роста зимних из-за изменений атмосферной циркуляции) соотносится с тенденциями глобального изменения климата. 

Это исследование можно использовать для локального мониторинга проявлений современного изменения климата, оценки темпов потепления и отслеживания изменений режима осадков на региональном уровне.
